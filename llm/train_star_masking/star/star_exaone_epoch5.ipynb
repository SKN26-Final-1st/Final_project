````python
# Cell 2. 기본 설정

import os
import re
import gc
import json
import random
import numpy as np
import pandas as pd
import torch

from pathlib import Path
from tqdm.auto import tqdm

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

# RunPod/Colab 환경에 맞게 수정
DATA_PATH = "/workspace/self_intro_star_dataset.jsonl"

# # 예시. 사용하려는 EXAONE 모델로 바꿔도 됨
# MODEL_ID = "LGAI-EXAONE/EXAONE-3.5-2.4B-Instruct"

# # 최신 main은 Transformers v5용 코드라 transformers 4.x와 충돌할 수 있음
# # 2024-12-11 commit: v5 업데이트 이전 revision
# MODEL_REVISION = "e949c91"

# Cell 2 쪽 설정에 추가

MODEL_ID = "LGAI-EXAONE/EXAONE-3.5-2.4B-Instruct"

# 이전에 정상 로드됐던 EXAONE revision
MODEL_REVISION = "e949c91dec92095908d34e6b560af77dd0c993f8"

WORK_DIR = Path("./exaone_self_intro_star")
WORK_DIR.mkdir(parents=True, exist_ok=True)

SPLIT_DIR = WORK_DIR / "splits"
SPLIT_DIR.mkdir(parents=True, exist_ok=True)

PRED_DIR = WORK_DIR / "predictions"
PRED_DIR.mkdir(parents=True, exist_ok=True)

ADAPTER_DIR = WORK_DIR / "exaone_star_lora"
````

````python
# Cell 3. JSONL 로드

def load_jsonl(path):
    rows = []
    with open(path, "r", encoding="utf-8-sig") as f:
        for i, line in enumerate(f, start=1):
            line = line.strip()
            if not line:
                continue
            try:
                rows.append(json.loads(line))
            except json.JSONDecodeError as e:
                raise ValueError(f"{i}번째 줄 JSON 파싱 실패: {e}")
    return rows

rows = load_jsonl(DATA_PATH)
df = pd.DataFrame(rows)

required_cols = {"label", "role", "input", "output"}
missing = required_cols - set(df.columns)
if missing:
    raise ValueError(f"필수 컬럼 누락: {missing}")

df["label"] = df["label"].astype(int)
df["role"] = df["role"].astype(str)
df["strata"] = df["label"].astype(str) + "||" + df["role"]

print("전체 데이터 수:", len(df))
print("label 분포")
display(df["label"].value_counts().sort_index())

print("role 수:", df["role"].nunique())
print("label+role strata 수:", df["strata"].nunique())
display(df.groupby(["label", "role"]).size().rename("count").reset_index().head())
````

````python
# Cell 4. label + role 기준 stratified 6:2:2 split
# sklearn train_test_split은 작은 strata에서 2차 split이 실패할 수 있어 직접 분배합니다.

def allocate_counts(n, ratios=(0.6, 0.2, 0.2)):
    ideal = np.array(ratios) * n
    counts = np.floor(ideal).astype(int)
    remain = n - counts.sum()

    order = np.argsort(-(ideal - counts))
    for idx in order[:remain]:
        counts[idx] += 1

    # strata 샘플이 3개 이상이면 train/val/test에 최소 1개씩 배치
    if n >= 3:
        for idx in range(3):
            if counts[idx] == 0:
                donor = int(np.argmax(counts))
                counts[donor] -= 1
                counts[idx] += 1

    return counts.tolist()  # train, val, test

def stratified_split_602020(df, strata_col="strata", seed=42):
    rng = np.random.default_rng(seed)

    train_idx, val_idx, test_idx = [], [], []

    small = df.groupby(strata_col).size()
    too_small = small[small < 3]
    if len(too_small) > 0:
        print("주의: 샘플이 3개 미만인 strata는 6:2:2를 완전히 보장할 수 없습니다.")
        display(too_small)

    for _, group in df.groupby(strata_col, sort=False):
        idx = group.index.to_numpy().copy()
        rng.shuffle(idx)

        n_train, n_val, n_test = allocate_counts(len(idx))

        train_idx.extend(idx[:n_train])
        val_idx.extend(idx[n_train:n_train+n_val])
        test_idx.extend(idx[n_train+n_val:n_train+n_val+n_test])

    train_df = df.loc[train_idx].sample(frac=1, random_state=seed).reset_index(drop=True)
    val_df = df.loc[val_idx].sample(frac=1, random_state=seed).reset_index(drop=True)
    test_df = df.loc[test_idx].sample(frac=1, random_state=seed).reset_index(drop=True)

    return train_df, val_df, test_df

train_df, val_df, test_df = stratified_split_602020(df, "strata", SEED)

print(len(train_df), len(val_df), len(test_df))
print("비율:", len(train_df)/len(df), len(val_df)/len(df), len(test_df)/len(df))

for name, part in [("train", train_df), ("val", val_df), ("test", test_df)]:
    part.drop(columns=["strata"]).to_json(
        SPLIT_DIR / f"{name}.jsonl",
        orient="records",
        lines=True,
        force_ascii=False,
    )

display(pd.DataFrame({
    "all": df["label"].value_counts(normalize=True).sort_index(),
    "train": train_df["label"].value_counts(normalize=True).sort_index(),
    "val": val_df["label"].value_counts(normalize=True).sort_index(),
    "test": test_df["label"].value_counts(normalize=True).sort_index(),
}))
````

````python
# Cell 5. 프롬프트 / 출력 포맷 함수

SYSTEM_PROMPT = """너는 자기소개서 답변을 STAR 구조로 변환하는 평가/작성 보조 모델이다.
반드시 JSON만 출력한다. 설명, 마크다운, 코드블록은 출력하지 않는다.
출력 key는 학습 데이터와 동일하게 self_intoduction, original_quality를 사용한다.
self_intoduction의 각 항목은 question과 answer를 포함하고, answer는 s, t, a, r을 포함한다."""

def build_messages(example):
    payload = {
        "role": example["role"],
        "input": example["input"],
    }

    user_prompt = f"""다음 JSON을 읽고 정답 output과 같은 구조의 JSON만 생성하라.

입력:
{json.dumps(payload, ensure_ascii=False)}

출력 JSON 스키마:
{{
  "self_intoduction": [
    {{
      "question": "...",
      "answer": {{
        "s": "...",
        "t": "...",
        "a": "...",
        "r": "..."
      }}
    }}
  ],
  "original_quality": "..."
}}"""

    return [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": user_prompt},
    ]

def target_text(example):
    return json.dumps(example["output"], ensure_ascii=False)
````

# base 평가용

````python
# Cell 6. 모델/토크나이저 로드 - base 평가용

from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig

def get_torch_dtype():
    if torch.cuda.is_available() and torch.cuda.is_bf16_supported():
        return torch.bfloat16
    return torch.float16

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=get_torch_dtype(),
    bnb_4bit_use_double_quant=True,
)

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_ID,
    revision=MODEL_REVISION,
    trust_remote_code=True,
    use_fast=True,
)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    revision=MODEL_REVISION,
    code_revision=MODEL_REVISION,
    quantization_config=bnb_config,
    device_map="auto",
    torch_dtype=get_torch_dtype(),
    trust_remote_code=True,
)

model.eval()
print("loaded:", MODEL_ID, MODEL_REVISION)
````

````python
# Cell 7. 미니배치 generation 함수

from pathlib import Path

tokenizer.padding_side = "left"

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

model.config.use_cache = True

def build_prompt(example):
    messages = build_messages(example)
    return tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
    )

def generate_batch(
    model,
    tokenizer,
    examples,
    max_new_tokens=800,
    max_input_tokens=2048,
):
    prompts = [build_prompt(ex) for ex in examples]

    inputs = tokenizer(
        prompts,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=max_input_tokens,
    ).to(model.device)

    with torch.inference_mode():
        output_ids = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            use_cache=True,
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.eos_token_id,
        )

    prompt_len = inputs["input_ids"].shape[1]
    generated = output_ids[:, prompt_len:]

    return [
        tokenizer.decode(ids, skip_special_tokens=True).strip()
        for ids in generated
    ]

def run_generation(
    model,
    tokenizer,
    data_df,
    output_path,
    batch_size=4,
    max_new_tokens=800,
    max_input_tokens=2048,
    resume=True,
):
    output_path = Path(output_path)
    records = []

    if resume and output_path.exists():
        with open(output_path, "r", encoding="utf-8") as f:
            records = [json.loads(line) for line in f if line.strip()]
        print(f"기존 예측 {len(records)}개 발견. 이어서 진행합니다.")

    start = len(records)

    with open(output_path, "a", encoding="utf-8") as f:
        for batch_start in tqdm(range(start, len(data_df), batch_size)):
            batch_df = data_df.iloc[batch_start:batch_start + batch_size]
            examples = [row.to_dict() for _, row in batch_df.iterrows()]

            preds = generate_batch(
                model=model,
                tokenizer=tokenizer,
                examples=examples,
                max_new_tokens=max_new_tokens,
                max_input_tokens=max_input_tokens,
            )

            for ex, pred in zip(examples, preds):
                record = {
                    "label": ex["label"],
                    "role": ex["role"],
                    "input": ex["input"],
                    "reference_output": ex["output"],
                    "prediction_text": pred,
                }
                records.append(record)
                f.write(json.dumps(record, ensure_ascii=False) + "\n")

            f.flush()

    return pd.DataFrame(records)
````

````python
# Cell 8. 평가 함수

from rouge_score import rouge_scorer
from sentence_transformers import SentenceTransformer

def extract_json_object(text):
    if not isinstance(text, str):
        return None

    text = text.strip()
    text = re.sub(r"^```json\s*", "", text)
    text = re.sub(r"^```\s*", "", text)
    text = re.sub(r"\s*```$", "", text)

    start = text.find("{")
    if start < 0:
        return None

    depth = 0
    in_str = False
    esc = False

    for i in range(start, len(text)):
        ch = text[i]

        if in_str:
            if esc:
                esc = False
            elif ch == "\\":
                esc = True
            elif ch == '"':
                in_str = False
        else:
            if ch == '"':
                in_str = True
            elif ch == "{":
                depth += 1
            elif ch == "}":
                depth -= 1
                if depth == 0:
                    return text[start:i+1]

    return None

def parse_prediction(text):
    obj_text = extract_json_object(text)
    if obj_text is None:
        return None
    try:
        return json.loads(obj_text)
    except Exception:
        return None

def schema_valid(obj):
    if not isinstance(obj, dict):
        return False
    if "self_intoduction" not in obj:
        return False
    if "original_quality" not in obj:
        return False
    if not isinstance(obj["self_intoduction"], list):
        return False

    for item in obj["self_intoduction"]:
        if not isinstance(item, dict):
            return False
        if "question" not in item or "answer" not in item:
            return False
        ans = item["answer"]
        if not isinstance(ans, dict):
            return False
        for k in ["s", "t", "a", "r"]:
            if k not in ans:
                return False

    return True

def flatten_output(obj):
    if not isinstance(obj, dict):
        return ""

    parts = []

    for item in obj.get("self_intoduction", []):
        if not isinstance(item, dict):
            continue

        parts.append(str(item.get("question", "")))

        ans = item.get("answer", {})
        if isinstance(ans, dict):
            for k in ["s", "t", "a", "r"]:
                parts.append(str(ans.get(k, "")))
        else:
            parts.append(str(ans))

    parts.append(str(obj.get("original_quality", "")))
    return "\n".join([p for p in parts if p])

def flatten_field(obj, field):
    if not isinstance(obj, dict):
        return ""

    parts = []

    if field == "original_quality":
        return str(obj.get("original_quality", ""))

    for item in obj.get("self_intoduction", []):
        ans = item.get("answer", {}) if isinstance(item, dict) else {}
        if isinstance(ans, dict):
            parts.append(str(ans.get(field, "")))

    return "\n".join([p for p in parts if p])

rouge = rouge_scorer.RougeScorer(["rougeL"], use_stemmer=False)

# 한국어/다국어 의미 유사도용. 더 큰 모델로 바꿔도 됨.
embedder = SentenceTransformer("sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")

def evaluate_prediction_df(pred_df):
    rows = []

    parsed_preds = [parse_prediction(x) for x in pred_df["prediction_text"]]
    ref_objs = pred_df["reference_output"].tolist()

    ref_texts = [flatten_output(x) for x in ref_objs]
    pred_texts = [flatten_output(x) if x is not None else "" for x in parsed_preds]

    ref_emb = embedder.encode(ref_texts, batch_size=16, normalize_embeddings=True, convert_to_numpy=True)
    pred_emb = embedder.encode(pred_texts, batch_size=16, normalize_embeddings=True, convert_to_numpy=True)
    semantic_sims = (ref_emb * pred_emb).sum(axis=1)

    for i, row in pred_df.reset_index(drop=True).iterrows():
        ref_obj = ref_objs[i]
        pred_obj = parsed_preds[i]

        ref_text = ref_texts[i]
        pred_text = pred_texts[i]

        rouge_l = rouge.score(ref_text, pred_text)["rougeL"].fmeasure if pred_text else 0.0

        metric_row = {
            "label": row["label"],
            "role": row["role"],
            "json_valid": pred_obj is not None,
            "schema_valid": schema_valid(pred_obj),
            "rougeL": rouge_l,
            "semantic_sim": float(semantic_sims[i]),
        }

        for field in ["s", "t", "a", "r", "original_quality"]:
            ref_field = flatten_field(ref_obj, field)
            pred_field = flatten_field(pred_obj, field) if pred_obj is not None else ""

            if ref_field and pred_field:
                f_ref = embedder.encode([ref_field], normalize_embeddings=True, convert_to_numpy=True)
                f_pred = embedder.encode([pred_field], normalize_embeddings=True, convert_to_numpy=True)
                metric_row[f"{field}_sim"] = float((f_ref * f_pred).sum())
            else:
                metric_row[f"{field}_sim"] = 0.0

        rows.append(metric_row)

    return pd.DataFrame(rows)

def summarize_eval(eval_df):
    metric_cols = [
        "json_valid", "schema_valid", "rougeL", "semantic_sim",
        "s_sim", "t_sim", "a_sim", "r_sim", "original_quality_sim",
    ]
    return eval_df[metric_cols].mean().to_frame("mean").T
````

````python
# Cell 11. SFT 학습 데이터 구성

from datasets import Dataset

def make_sft_text(example):
    messages = build_messages(example)
    messages.append({
        "role": "assistant",
        "content": target_text(example),
    })

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=False,
    )

    if tokenizer.eos_token and not text.endswith(tokenizer.eos_token):
        text += tokenizer.eos_token

    return text

train_records = train_df.drop(columns=["strata"]).to_dict("records")
val_records = val_df.drop(columns=["strata"]).to_dict("records")

train_sft = Dataset.from_list([{"text": make_sft_text(x)} for x in train_records])
val_sft = Dataset.from_list([{"text": make_sft_text(x)} for x in val_records])

print(train_sft[0]["text"][:1000])
````

````python
# Cell 12. QLoRA SFT 학습 - early stopping + best checkpoint 저장

from peft import LoraConfig
from trl import SFTTrainer, SFTConfig
from transformers import EarlyStoppingCallback

model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    revision=MODEL_REVISION,
    code_revision=MODEL_REVISION,
    quantization_config=bnb_config,
    device_map="auto",
    torch_dtype=get_torch_dtype(),
    trust_remote_code=True,
)

model.config.use_cache = False

def guess_lora_targets(model):
    candidates = {
        "q_proj", "k_proj", "v_proj", "o_proj",
        "gate_proj", "up_proj", "down_proj",
    }
    found = sorted({
        name.split(".")[-1]
        for name, module in model.named_modules()
        if name.split(".")[-1] in candidates
    })
    return found or sorted(candidates)

lora_config = LoraConfig(
    r=16,
    lora_alpha=32,
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
    target_modules=guess_lora_targets(model),
)

# train 약 300개 기준:
# batch 1 * grad_acc 8 => optimizer step 약 38 step / epoch
# eval_steps=20이면 epoch당 약 2번 검증
# patience=3이면 eval loss가 3번 연속 좋아지지 않을 때 조기 종료
sft_args = SFTConfig(
    output_dir=str(ADAPTER_DIR),
    dataset_text_field="text",
    max_length=2048,
    packing=False,

    num_train_epochs=5,
    per_device_train_batch_size=1,
    per_device_eval_batch_size=1,
    gradient_accumulation_steps=8,

    learning_rate=2e-4,
    warmup_ratio=0.03,
    lr_scheduler_type="cosine",

    logging_steps=5,

    eval_strategy="steps",
    eval_steps=20,

    save_strategy="steps",
    save_steps=20,
    save_total_limit=3,

    load_best_model_at_end=True,
    metric_for_best_model="eval_loss",
    greater_is_better=False,

    bf16=torch.cuda.is_available() and torch.cuda.is_bf16_supported(),
    fp16=torch.cuda.is_available() and not torch.cuda.is_bf16_supported(),

    report_to="none",
    seed=SEED,
)

early_stopping = EarlyStoppingCallback(
    early_stopping_patience=3,
    early_stopping_threshold=0.001,
)

trainer = SFTTrainer(
    model=model,
    args=sft_args,
    train_dataset=train_sft,
    eval_dataset=val_sft,
    peft_config=lora_config,
    processing_class=tokenizer,
    callbacks=[early_stopping],
)

train_result = trainer.train()

print("Best checkpoint:", trainer.state.best_model_checkpoint)
print("Best eval loss:", trainer.state.best_metric)

# load_best_model_at_end=True 이므로 여기서 저장되는 모델은 best checkpoint 기준 adapter
trainer.save_model(str(ADAPTER_DIR))
tokenizer.save_pretrained(str(ADAPTER_DIR))

print("Best LoRA adapter saved to:", ADAPTER_DIR)
````
